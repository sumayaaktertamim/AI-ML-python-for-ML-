```python
# prblm solve-2:positive,negative,even,odd
n=int(input())
inp=input()
nums=inp.split()
positive=0 ; negative=0 ; even=0; odd=0
for i in range(n):
    x=int(nums[i])
    # positive or negative
    if x>0:
        positive+=1
    elif x<0:
        negative+=1
    # even or odd 
    if x%2 !=0:
        odd+=1
    else:
        even+=1
print('Even:', even)
print("Odd:", odd )
print('Positive:', positive)
print('Negative:', negative)
        
```

```python
# if else
raining=False
if raining==True:
    print('Rain coat is nessary')
else:
    print('Go without a Rain coat')
    
```

```python
# Nested if else
raining=True
Muddy=False
if raining==True:
    print('Rain coat is nessary')
    if Muddy==True:
        print('Sandle is nessary')
    else:
        print('Bring Sneakers')
else:
    print('Go without a Rain coat')
```
